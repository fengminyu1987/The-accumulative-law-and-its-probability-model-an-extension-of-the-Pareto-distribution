```python
import numpy as np
```

```python
# website visit distribution   Eq. (4.16)
def p_3(T, beta):
    return (1 - np.exp(-1 * beta * T) - beta * T *  np.exp(-1 * beta * T)) / (beta * T**2)
```

```python
beta = 1.90
T_list = np.arange(1, 151)
result = []
for T in T_list:
    result.append(p_3(T, beta))

result_new = []
for i in result:
    result_new.append(i / sum(result))
```
