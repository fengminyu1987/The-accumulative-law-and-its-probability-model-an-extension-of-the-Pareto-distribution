```python
import numpy as np
from scipy import integrate
```

```python
# degree distribution   Eq. (4.6)
def f_1(x, mu, sigma):
    return x * np.exp(-1 * (np.log(x) - mu)**2 / (2 * sigma**2)) / (np.sqrt(2 * np.pi) * sigma)
def p_1(k, mu, sigma):
    temp = integrate.fixed_quad(lambda x: f_1(x, mu, sigma), 0, k, n = 100)[0]
    return 2 * temp / k**3
```

```python
mu = 1
sigma = 0.5
k_list = np.arange(0, 51)
result = []
for k in k_list:
    if k == 0:
        result.append(0)
    else:
        result.append(p_1(k, mu, sigma))
```
