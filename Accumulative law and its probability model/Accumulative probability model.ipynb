```python
import numpy as np
from scipy import integrate
```

```python
# Accumulative probability model   Eq. (2.18)
def f_1(x, mu, sigma, alpha):
    return x**(alpha - 1) * np.exp(-1 * (np.log(x) - mu)**2 / (2 * sigma**2)) / (np.sqrt(2 * np.pi) * sigma)
def p_1(z, mu, sigma, alpha):
    temp = integrate.fixed_quad(lambda x: f_1(x, mu, sigma, alpha), 0, z, n = 100)[0]
    return alpha * temp / z**(alpha + 1)
```

```python
mu = float(input("Please enter the value of mu: "))
sigma = float(input("Please enter the value of sigma: "))
alpha = float(input("Please enter the value of alpha: "))
z = float(input("Please enter the value of z: "))
if z == 0:
    print('The result is:', 0)
else:
    print('The result is:', p_1(z, mu, sigma, alpha))
```
