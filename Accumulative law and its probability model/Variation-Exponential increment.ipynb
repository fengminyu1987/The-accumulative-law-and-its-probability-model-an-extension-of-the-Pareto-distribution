```python
import numpy as np
```

```python
# Variation: Exponential increment   Eq. (2.22)
def p_3(z, alpha, beta):
    return alpha * (1 - np.exp(-1 * beta * z) - beta * z * np.exp(-1 * beta * z)) / (beta * z**(alpha + 1))
```

```python
alpha = float(input("Please enter the value of alpha: "))
beta = float(input("Please enter the value of beta: "))
z = float(input("Please enter the value of z: "))
print('The result is:', p_3(z, alpha, beta))
```
