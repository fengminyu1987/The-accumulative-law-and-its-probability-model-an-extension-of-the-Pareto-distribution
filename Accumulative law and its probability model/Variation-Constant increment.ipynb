```python
# Variation: Constant increment   Eq. (2.20)
def p_2(z, alpha, y):
    return alpha * y**alpha / z**(alpha + 1)
```

```python
alpha = float(input("Please enter the value of alpha: "))
y = float(input("Please enter the value of y: "))
z = float(input("Please enter the value of z: "))
print('The result is:', p_2(z, alpha, y))
```
