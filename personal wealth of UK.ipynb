```python
import numpy as np
from scipy import integrate
```

```python
# personal wealth distribution   Eq. (4.11)
def f_2(x, mu, sigma):
    return np.exp(-1 * (np.log(x) - mu)**2 / (2 * sigma**2)) / (np.sqrt(2 * np.pi) * sigma)
def p_2(w, mu, sigma):
    temp = integrate.fixed_quad(lambda x: f_2(x, mu, sigma), 0, w, n = 100)[0]
    return temp / w**2
```

```python
mu = 1.00
sigma = 0.78
w_list = [0, 1, 2, 3, 5, 10, 20, 30]
result = []
for w in w_list:
    if w == 0:
        result.append(0)
    else:
        result.append(p_2(w, mu, sigma))
        
result_new = []
for i in result:
    result_new.append(i / sum(result))
```
