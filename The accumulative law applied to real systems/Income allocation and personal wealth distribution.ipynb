```python
import numpy as np
from scipy import integrate
```

```python
# personal wealth distribution   Eq. (4.11)
def f_2(x, mu, sigma):
    return np.exp(-1 * (np.log(x) - mu)**2 / (2 * sigma**2)) / (np.sqrt(2 * np.pi) * sigma)
def p_2(w, mu, sigma):
    temp = integrate.fixed_quad(lambda x: f_2(x, mu, sigma), 0, w, n = 100)[0]
    return temp / w**2
```

```python
mu = float(input("Please enter the value of mu: "))
sigma = float(input("Please enter the value of sigma: "))
w = float(input("Please enter the value of w: "))
if w == 0:
    print('The result is:', 0)
else:
    print('The result is:', p_2(w, mu, sigma))
```
