```python
import numpy as np
```

```python
# website visit distribution   Eq. (4.16)
def p_3(T, beta):
    return (1 - np.exp(-1 * beta * T) - beta * T * np.exp(-1 * beta * T)) / (beta * T**2)
```

```python
beta = float(input("Please enter the value of beta: "))
T = float(input("Please enter the value of T: "))
print('The result is:', p_3(T, beta))
```
